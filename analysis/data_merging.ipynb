## Removing duplicate data

I just want to add raw_data_1 onto the end of raw_data

That will involve getting rid of the first few rows of raw_data_1 as they are duplicates

```python
data_from_2012.head(20)
```

```python
data_from_2020.tail(20)
```

So I need to remove the last 15 rows of the first df

```python
# Number of rows to drop
n = 15

# Dropping last n rows using drop
data_from_2020.drop(data_from_2020.tail(n).index,
        inplace = True)
```

```python
# checking it worked
data_from_2020.tail(20)
```

## Merging the datasets

```python
master_list = pd.concat([data_from_2020, data_from_2012])
```

```python
master_list.head()
```

```python
master_list.tail()
```

```python
master_list["description"]
```

```python
master_list.info()
```

```python
# Setting the correct index
master_list.index = range(13795)
```

```python
# Converting dates to datetime
master_list["published_date_as_date"] = pd.to_datetime(master_list["published_date"])
master_list["bestsellers_date_as_date"] = pd.to_datetime(master_list["bestsellers_date"])
```

```python
master_list.to_csv("raw_data/raw_data_merged.csv")
```

## Subsetting data for the last decade 

```python
df = master_list[master_list.bestsellers_date_as_date > "2016-01-01"]
```

```python
df.info()
```

```python
# removign any that don't have a description 
df.dropna(subset=["description"], inplace=True)
```

```python
df.to_csv("data_for_analysis/bestsellers_decade_data.csv")
```

```python
df.info()
```

## Creating a subset of unique values for the AI to analyse

This will keep the cost down and make the analysis run faster as the AI will only need to do the classification on a subset of the data

```python
unique_df = df.drop_duplicates(subset="title")
```
